```python
import heapq



start = [1, 2, 3,
         0, 4, 6,
         7, 5, 8]


goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]




def heuristic(state):

    h = 0

    for i in range(9):

        if state[i] == 0:
            continue

        goal_position = goal.index(state[i])

        row1 = i // 3
        col1 = i % 3

        row2 = goal_position // 3
        col2 = goal_position % 3

        h = h + abs(row1 - row2) + abs(col1 - col2)

    return h




def get_moves(state):

    moves = []

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    if row > 0:
        new_state = state.copy()
        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]

        moves.append(new_state)

    # DOWN
    if row < 2:
        new_state = state.copy()
        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]

        moves.append(new_state)

    # LEFT
    if col > 0:
        new_state = state.copy()
        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]

        moves.append(new_state)

    # RIGHT
    if col < 2:
        new_state = state.copy()
        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]

        moves.append(new_state)

    return moves


def show(state):

    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])



def astar():

    # (F, G, state, path)
    queue = []

    heapq.heappush(
        queue,
        (heuristic(start), 0, start, [start])
    )

    visited = set()

    while queue:

        f, g, state, path = heapq.heappop(queue)

        state_tuple = tuple(state)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)

        # GOAL FOUND
        if state == goal:
            return path

        # Generate next states
        for next_state in get_moves(state):

            if tuple(next_state) not in visited:

                new_g = g + 1
                new_h = heuristic(next_state)
                new_f = new_g + new_h

                new_path = path + [next_state]

                heapq.heappush(
                    queue,
                    (new_f, new_g, next_state, new_path)
                )

    return None




print("START:")
show(start)

print("\nGOAL:")
show(goal)

path = astar()



if path is None:

    print("\nNo solution possible.")

else:

    print("\nSOLUTION:")

    for i in range(len(path)):

        print("\nStep", i)
        show(path[i])

    print("\nGoal reached!")
    print("Total moves =", len(path) - 1)
```
